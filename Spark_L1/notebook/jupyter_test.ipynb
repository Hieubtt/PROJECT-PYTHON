```python
from pyspark.sql import SparkSession
import pandas as pd
import numpy as np
import random

data = {
    "id_du_an": range(1, 501),
    "ten_du_an": [f"Cong trinh {i}" for i in range(1, 501)],
    "loai_vat_lieu": [random.choice(["Thep", "Xi mang", "Cat", "Da", "Gach"]) for _ in range(500)],
    "nha_thau": [random.choice(["Delta", "Coteccons", "Hoa Binh", "Vinaconex", "An Phong"]) for _ in range(500)],
    "gia_tri_goi_thau": np.random.randint(1000, 50000, 500), # Đơn vị: Triệu VNĐ
    "so_luong_vat_lieu": np.random.randint(100, 10000, 500),
    "don_gia": np.random.randint(10, 500, 500),
    "ngay_bat_dau": pd.to_datetime(['2024-01-01'] * 500) + pd.to_timedelta(np.random.randint(0, 365, 500), unit='D'),
    "trang_thai": [random.choice(["Dang thi cong", "Hoan thanh", "Tam dung", "Cho duyet"]) for _ in range(500)],
    "khu_vuc": [random.choice(["Mien Bac", "Mien Trung", "Mien Nam"]) for _ in range(500)]
}
pdf = pd.DataFrame(data)

# Chuyển từ Pandas DataFrame sang Spark DataFrame
df = spark.createDataFrame(pdf)

# Hiển thị 5 dòng đầu tiên
df.show(5)
```
